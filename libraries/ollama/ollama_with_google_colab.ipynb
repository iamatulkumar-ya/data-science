```python
# source code: https://github.com/ollama/ollama-python
from ollama import chat
from ollama import ChatResponse

def get_model_response(modelName:str, contentText:str) -> str:

  response: ChatResponse = chat(model=modelName, messages=[
    {
      'role': 'user',
      'content': contentText,
    },
  ])
  # print(response['message']['content'])
  # or access fields directly from the response object
  return response.message.content
```

```python
# testing with llama3.2:1b
print(get_model_response('llama3.2:1b', "Whay sky is blue?"))
```

```python
# testing with gemma3:270m
print(get_model_response('llama3.2:1b', "Whay sky is blue?"))
```
